# Product Recommendation Model
This notebook implements a Collaborative Filtering recommendation engine using the SVD (Singular Value Decomposition) algorithm.

```python
!pip install "numpy<2" scikit-surprise
```

```python
import pandas as pd
from surprise import SVD, Dataset, Reader, accuracy
from surprise.model_selection import train_test_split

# Load the MovieLens 100k dataset (standard benchmark for recommendation)
data = Dataset.load_builtin('ml-100k')

# Convert to a pandas dataframe for inspection
df = pd.DataFrame(data.raw_ratings, columns=['user', 'item', 'rating', 'timestamp'])
display(df.head())
```

### Train the Recommendation Model
We will split the data into training and testing sets, then use the SVD algorithm to learn user-item interactions.

```python
# Split into train and test set (75% train, 25% test)
trainset, testset = train_test_split(data, test_size=0.25)

# Use the SVD algorithm
algo = SVD()

# Train the algorithm on the trainset
algo.fit(trainset)

# Predict ratings for the testset
predictions = algo.test(testset)

# Compute RMSE (Root Mean Squared Error)
accuracy.rmse(predictions)
```

### Predict a specific rating
Let's predict what rating User '196' would give to Item '302'.

```python
uid = str(196)
iid = str(302)

pred = algo.predict(uid, iid, verbose=True)
print(f"Predicted rating for user {uid} on item {iid} is: {pred.est:.2f}")
```
